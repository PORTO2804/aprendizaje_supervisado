## 4. Qué produjo la IA en cada etapa

```python
import os, json
import pandas as pd

cuadernos = sorted((f for f in os.listdir(".") if f.endswith(".ipynb") and f[0].isdigit()), key=lambda f: int(f.split("_")[0]))
filas = []
for f in cuadernos:
    nb = json.load(open(f, encoding="utf-8"))
    code = [c for c in nb["cells"] if c["cell_type"] == "code"]
    md = [c for c in nb["cells"] if c["cell_type"] == "markdown"]
    lineas = sum(len("".join(c["source"]).splitlines()) for c in code)
    errores = sum(1 for c in code for o in c.get("outputs", []) if o.get("output_type") == "error")
    filas.append({"cuaderno": f, "celdas_codigo": len(code), "celdas_texto": len(md),
                  "lineas_codigo": lineas, "errores_en_salida": errores})
resumen = pd.DataFrame(filas).set_index("cuaderno")
for extra in ["titanic_features.py", "app.py"]:
    resumen.loc[extra, "lineas_codigo"] = len(open(extra, encoding="utf-8").read().splitlines())
resumen = resumen.fillna(0).astype(int)
resumen.loc["TOTAL"] = resumen.sum()
resumen
```

## 7. Reflexión sobre el uso de IA generativa en este proyecto

### Lo que aportó
- **Velocidad:** 10 cuadernos, un módulo reutilizable y una app en ~2 horas, incluidas pruebas estadísticas avanzadas (Nadeau-Bengio, Holm, SHAP).
- **Profundidad:** la IA propuso análisis que no estaban en la consigna original (validación repetida, umbral, SHAP, fuga) y que fortalecieron la justificación del modelo.
- **Autocrítica cuando se le pide:** al preguntarle "¿qué recomiendas?", señaló sus propios problemas (contradicción del cuaderno 4, fuga de información).

### Riesgos observados
- **Cifras inventadas pero plausibles** (error 5): un "~38 %" que en realidad era 34.4 %. Si no se ejecuta y se compara, pasa desapercibido.
- **Hipótesis convincentes pero falsas** (errores 9–11): las explicaciones de la IA sonaban lógicas y los datos las desmintieron. **Una explicación bien redactada no es evidencia.**
- **Errores silenciosos** (error 4): el código corría sin fallar y aun así el resultado era incorrecto.
- **Dependencia de versiones** (error 3): el código generado puede romper archivos guardados antes.

### Buenas prácticas que funcionaron
1. **Dar objetivos completos al inicio** (lista de modelos y métricas).
2. **Trabajar por etapas y revisar cada una** antes de seguir.
3. **Exigir ejecución real**, no solo código.
4. **Pedir una simulación desde cero** al final para confirmar que todo es reproducible.
5. **Pedirle a la IA que diga sus dudas antes de ejecutar** (etapa 7) para no asumir el formato.

### Conclusión
La IA generativa fue eficaz para **escribir y ejecutar** el código de un proyecto completo de aprendizaje supervisado. El valor del trabajo, sin embargo, dependió de **verificar todo contra los resultados reales**: 11 de los 13 problemas registrados (todos menos los 2 obstáculos de herramientas) se detectaron porque el código se ejecutó y las salidas se revisaron. La IA escribe con mucha seguridad incluso cuando se equivoca, por lo que **la responsabilidad de validar sigue siendo humana**.
